# Trilha Secundária: Chem-E-Car — Módulo 02: Projeto de Calibração

Desafio prático focado na modelagem do comportamento do carrinho para a competição:
- Ajuste de curva linear de calibração usando `numpy`
- Construção de um algoritmo preditivo de massa de reagente
- Avaliação do erro percentual de parada em relação à distância alvo

## 1. Carregamento dos Dados e Regressão Linear
Ajuste da equação de reta $Y = aX + b$ para mapear a relação entre massa e distância.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/isaulfelipe/miniCursoPython/main/trilha-secundaria-chem-e-car/dados/testes_carrinho.csv"
df = pd.read_csv(url)

# Obtendo os coeficientes da reta de calibração (a = inclinação, b = intercepto)
a, b = np.polyfit(df['distancia_percorrida_m'], df['massa_reagente_g'], 1)
print(f"Equação de Calibração: Massa(g) = {a:.4f} * Distância(m) + {b:.4f}")
```

## 2. Calculadora de Carga Útil
Função preditiva para uso no dia da competição.

```python
def predizer_massa_necessaria(distancia_alvo_m, coef_a, coef_b):
    """Retorna a massa de reagente necessária para atingir a distância alvo em metros."""
    return (coef_a * distancia_alvo_m) + coef_b

# Exemplo: Distância sorteada na competição = 12.5 metros
distancia_sorteada = 12.5
massa_calculada = predizer_massa_necessaria(distancia_sorteada, a, b)
print(f"Massa recomendada para {distancia_sorteada}m: {massa_calculada:.2f} g")
```

## 3. Curva de Calibração com a Reta de Regressão
Plotagem dos pontos experimentais junto com a linha de tendência.

```python
x_linha = np.linspace(df['distancia_percorrida_m'].min(), df['distancia_percorrida_m'].max(), 100)
y_linha = a * x_linha + b

plt.figure(figsize=(9, 5))
plt.scatter(df['distancia_percorrida_m'], df['massa_reagente_g'], color='red', label='Medições Reais')
plt.plot(x_linha, y_linha, color='blue', linestyle='--', label='Modelo Linear (Calibração)')
plt.title('Modelo de Calibração Chem-E-Car')
plt.xlabel('Distância Percorrida (m)')
plt.ylabel('Massa de Reagente (g)')
plt.grid(True)
plt.legend()
plt.show()
```
