## 1. Importação das Bibliotecas e Leitura dos Dados
Utilizaremos a biblioteca `pandas` para manipular tabelas e o `matplotlib` para visualização.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Carregando os dados de testes experimentais do repositório
url = "https://raw.githubusercontent.com/isaulfelipe/miniCursoPython/main/trilha-secundaria-chem-e-car/dados/testes_carrinho.csv"
df = pd.read_csv(url)

# Exibindo as primeiras linhas da tabela
df.head()
```

## 2. Inspeção e Estatística Básica
Analisando métricas gerais do conjunto de dados.

```python
# Resumo estatístico (média, desvio padrão, min, max)
print("Resumo Estatístico:")
print(df.describe())

# Filtrando apenas os testes onde a distância percorrida foi maior que 10 metros
testes_longos = df[df['distancia_percorrida_m'] > 10.0]
print("\nTestes com distância > 10m:")
print(testes_longos[['teste_id', 'massa_reagente_g', 'distancia_percorrida_m']])
```

## 3. Visualização de Dados
Gerando gráfico de **Massa de Reagente vs. Distância Percorrida**.

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['massa_reagente_g'], df['distancia_percorrida_m'], color='blue', label='Testes Reais')
plt.title('Calibração do Carrinho: Massa x Distância')
plt.xlabel('Massa do Reagente (g)')
plt.ylabel('Distância Percorrida (m)')
plt.grid(True)
plt.legend()
plt.show()
```
