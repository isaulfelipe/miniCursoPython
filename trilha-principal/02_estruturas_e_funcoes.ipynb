## 1. Estruturas Condicionais
Tomada de decisão com base em expressões lógicas.

```python
pressao = float(input("Digite a pressão do sistema (bar): "))

if pressao > 10.0:
    print("CRÍTICO: Pressão acima do limite seguro!")
elif pressao >= 5.0:
    print("ALERTA: Pressão em nível moderado.")
else:
    print("NORMAL: Pressão em nível seguro.")
```

## 2. Listas e Laço `for`
Listas armazenam conjuntos ordenados de dados. O laço `for` percorre esses elementos.

```python
# Lista de medições de tempo de parada (em segundos)
tempos = [12.4, 11.8, 12.1, 13.0, 11.9]

# Iterando sobre a lista
soma = 0
for t in tempos:
    soma += t

media = soma / len(tempos)
print(f"Total de testes: {len(tempos)}")
print(f"Tempo médio de parada: {media:.2f} s")
```

## 3. Funções (`def`)
Blocos de código reutilizáveis que recebem parâmetros e retornam valores.

```python
def calcular_massa_reagente(distancia_alvo_m):
    """Calcula a massa aproximada de reagente baseada na distância alvo."""
    # Equação linear hipotética: Y = 1.5 * X + 2.0
    massa_g = (1.5 * distancia_alvo_m) + 2.0
    return massa_g

# Testando a função
distancia = 10.0  # metros
massa_necessaria = calcular_massa_reagente(distancia)
print(f"Para percorrer {distancia} m, serão necessários {massa_necessaria} g de reagente.")
```

## ✏️ Exercício Prático 02
Crie uma função chamada `verificar_desvio` que recebe a distância esperada e a distância real. Se a diferença absoluta for maior que 0.5 metros, exiba uma mensagem alertando que a parada precisa de calibração.

```python
# Escreva seu código abaixo:
def verificar_desvio(esperada, real):
    desvio = abs(esperada - real)
    if desvio > 0.5:
        print(f"Calibração necessária! Desvio de {desvio:.2f} m.")
    else:
        print(f"Parada aceitável. Desvio de {desvio:.2f} m.")

verificar_desvio(10.0, 10.8)
```
