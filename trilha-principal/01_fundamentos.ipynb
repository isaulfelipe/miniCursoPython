## 1. Saída de Dados e Variáveis
A função `print()` exibe mensagens na tela. Para armazenar informações, usamos variáveis.

```python
# Exibindo uma mensagem
print("Olá, Capítulo AIChE!")

# Declarando variáveis de diferentes tipos
nome_membro = "Ana"          # String (str)
idade = 20                   # Inteiro (int)
altura = 1.68                # Ponto Flutuante (float)
ativo = True                 # Booleano (bool)

print(f"Membro: {nome_membro} | Idade: {idade} | Altura: {altura}m")
```

## 2. Entrada de Dados (`input`)
A função `input()` recebe dados digitados pelo usuário (sempre no formato texto/string).

```python
# Capturando entrada
curso = input("Digite o seu curso: ")
print(f"Bem-vindo(a) ao minicurso, estudante de {curso}!")

# Conversão de tipo (Casting)
massa_str = input("Digite a massa do reagente em gramas: ")
massa = float(massa_str)  # Converte a string para float
print(f"Massa registrada: {massa} g (Tipo: {type(massa)})")
```

## 3. Operadores Aritméticos
Operações matemáticas básicas em Python:

```python
a = 10
b = 3

print(f"Soma: {a + b}")
print(f"Subtração: {a - b}")
print(f"Multiplicação: {a * b}")
print(f"Divisão: {a / b}")
print(f"Divisão Inteira: {a // b}")
print(f"Resto da Divisão: {a % b}")
print(f"Potenciação: {a ** b}")
```

## ✏️ Exercício Prático 01
Escreva um programa que peça a temperatura em Celsius e converta para Kelvin ($K = C + 273.15$).

```python
# Escreva seu código abaixo:
celsius = float(input("Temperatura em °C: "))
kelvin = celsius + 273.15
print(f"Temperatura equivalente: {kelvin} K")
```
